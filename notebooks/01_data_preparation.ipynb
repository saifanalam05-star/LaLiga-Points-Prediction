```python
# Pandas is used to load, clean, and manipulate datasets
import pandas as pd

# NumPy is used for numerical operations and missing values
import numpy as np
```

```python
# STEP 1: ADD THE 2025/2026 SEASON TO OUR EXISTING DATASET

# Load our original La Liga dataset (2014/15 to 2024/25)
laliga = pd.read_csv("laligamatches.csv")

# Load the new 2025/26 season
new_season = pd.read_csv("laliga_2025_2026.csv")

# Standardize column names to match our original dataset
new_season.columns = (
    new_season.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

# Identify the season using its starting year
new_season["season"] = 2025

# Add missing columns and match the original column order
# Missing xG values will automatically become NaN
new_season = new_season.reindex(columns=laliga.columns)

# Replace any existing 2025/26 rows to avoid duplicates
laliga = laliga[laliga["season"] != 2025]

# Combine the original dataset with the new season
laliga_updated = pd.concat(
    [laliga, new_season],
    ignore_index=True
)

# Verify that we have 20 teams for the new season
assert len(new_season) == 20, "Expected 20 teams for 2025/26"

# Verify that no season/team combination is duplicated
assert not laliga_updated.duplicated(
    subset=["season", "squad"]
).any(), "Duplicate season/team detected"

# Save the updated dataset back to our original CSV
laliga_updated.to_csv("laligamatches.csv", index=False)

# Display the results
print("Dataset successfully updated!")
print("Total rows and columns:", laliga_updated.shape)
print("Number of seasons:", laliga_updated["season"].nunique())

# Display the newly added season
display(laliga_updated[laliga_updated["season"] == 2025].head())
```

```python
# Inspect our combined dataset

# Load the combined La Liga dataset
laliga = pd.read_csv("laligamatches.csv")

# Display the dataset dimensions
print("Dataset shape:", laliga.shape)

# Display information about each column
# This includes data types and non-null values
print("\nDataset information:")
laliga.info()

# Count missing values in each column
print("\nMissing values per column:")
print(laliga.isnull().sum())

# Display the first five rows
display(laliga.head())
```

```python
# STEP 3: CHECK MISSING VALUES BY SEASON

# Group the dataset by season
# Count how many missing values each season has
missing_by_season = (
    laliga.groupby("season")
    [["attendance", "xg", "xga", "xgd", "xgd/90"]]
    .apply(lambda group: group.isnull().sum())
)

# Display the results
display(missing_by_season)
```

```python
# STEP 4: VALIDATE OUR FOOTBALL STATISTICS

# Check that wins + draws + losses = matches played
invalid_matches = laliga[
    laliga["w"] + laliga["d"] + laliga["l"] != laliga["mp"]
]

# Check that goals for - goals against = goal difference
invalid_goal_difference = laliga[
    laliga["gf"] - laliga["ga"] != laliga["gd"]
]

# Check that points = 3 * wins + draws
invalid_points = laliga[
    3 * laliga["w"] + laliga["d"] != laliga["pts"]
]

# Check for duplicate teams within the same season
duplicates = laliga[
    laliga.duplicated(subset=["season", "squad"], keep=False)
]

# Display validation results
print("Invalid match totals:", len(invalid_matches))
print("Invalid goal differences:", len(invalid_goal_difference))
print("Invalid point totals:", len(invalid_points))
print("Duplicate team-season records:", len(duplicates))
```

```python
# STEP 5: INVESTIGATE INVALID POINT TOTALS

# Display the team whose points don't match
# the standard football points formula
display(
    invalid_points[
        ["season", "squad", "mp", "w", "d", "l", "pts", "notes"]
    ]
)

# Calculate how many points the team should have
# based on its wins and draws
invalid_points = invalid_points.copy()

invalid_points["expected_pts"] = (
    3 * invalid_points["w"] + invalid_points["d"]
)

# Calculate the difference between actual and expected points
invalid_points["difference"] = (
    invalid_points["pts"] - invalid_points["expected_pts"]
)

# Display the comparison
display(
    invalid_points[
        ["season", "squad", "pts", "expected_pts", "difference"]
    ]
)
```

```python
# STEP 6: SAVE OUR CLEANED DATASET

# Create a separate copy of our validated dataset
# This preserves the original DataFrame
laliga_cleaned = laliga.copy()

# Sort teams chronologically by season and league position
laliga_cleaned = laliga_cleaned.sort_values(
    by=["season", "rk"]
).reset_index(drop=True)

# Save the cleaned dataset to a new CSV file
# index=False prevents pandas from adding an extra index column
laliga_cleaned.to_csv("laliga_cleaned.csv", index=False)

# Verify that the file was saved correctly
print("Cleaned dataset saved successfully!")
print("Dataset shape:", laliga_cleaned.shape)
print("Seasons:", laliga_cleaned["season"].nunique())

# Display the last five rows
display(laliga_cleaned.tail())
```

```python
# STEP 7: INSPECT TEAM NAME CONSISTENCY

# Display all unique team names in alphabetical order
team_names = sorted(laliga_cleaned["squad"].unique())

print("Number of unique team names:", len(team_names))

# Display the team names
for team in team_names:
    print(team)
```

```python
# STEP 8: STANDARDIZE TEAM NAMES

# Create a dictionary mapping inconsistent names
# to a single standardized team name
team_name_mapping = {
    "Real Betis": "Betis"
}

# Replace inconsistent team names
laliga_cleaned["squad"] = (
    laliga_cleaned["squad"].replace(team_name_mapping)
)

# Check the number of unique teams
print("Unique teams:", laliga_cleaned["squad"].nunique())

# Verify that Real Betis is now consistently named Betis
display(
    laliga_cleaned[
        laliga_cleaned["squad"] == "Betis"
    ][["season", "squad", "pts"]]
    .sort_values("season")
)

# Check for duplicate team-season records
print(
    "Duplicate records:",
    laliga_cleaned.duplicated(
        subset=["season", "squad"]
    ).sum()
)
```

```python
# STEP 9: SAVE CORRECTED DATASET

# Save the updated cleaned dataset
# without modifying the original raw CSV
laliga_cleaned.to_csv(
    "../LinearRegression/laliga_cleaned.csv",
    index=False
)

print("Corrected dataset saved successfully!")
print("Dataset shape:", laliga_cleaned.shape)
```
